**Found missing values inside column 'Review Rating' and filled using median or mean . Here i used median because it is most robust to outliers**

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.groupby('Category')['Review Rating'].median()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df.columns
```

```python
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
df['age_group'] = pd.qcut(
    df['age'],
    q=4,
    labels=['young_adult', 'adult', 'middle_aged', 'senior']

)
```

```python
df.columns.name = None
df['age_group']
```

```python
frequency_days = {
    'Weekly': 7,
    'Fortnightly': 14,
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Every 3 Months': 90,
    'Annually': 365
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_days)
```
